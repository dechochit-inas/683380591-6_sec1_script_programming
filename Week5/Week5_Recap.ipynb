## 1. Dictionaries (The Key-Value Store)

Dictionary คือโครงสร้างข้อมูลที่เก็บข้อมูลเป็นคู่แบบ Key-Value โดยสามารถเปลี่ยนแปลงค่าได้ (Mutable)

* **Key** จะต้องไม่ซ้ำกันและต้องเป็นชนิดข้อมูลที่ไม่สามารถเปลี่ยนแปลงได้ (Immutable) เช่น String, Number หรือ Tuple


* **ประสิทธิภาพขั้นสูง:** ใน Python นั้น Dictionary ถูกสร้างขึ้นบนสถาปัตยกรรม Hash Tables ทำให้ความเร็วในการค้นหาข้อมูล (Lookup speed) คงที่ เฉลี่ยอยู่ที่ O(1) เสมอ ไม่ว่าข้อมูลจะมีขนาดใหญ่แค่ไหนก็ตาม


* ตั้งแต่ Python เวอร์ชัน 3.7 เป็นต้นมา Dictionary จะจดจำลำดับการเพิ่มข้อมูล (Insertion order) ไว้เป็นค่าเริ่มต้น



### Code Snippet: การเข้าถึงข้อมูลอย่างปลอดภัย (Safe Access)

การใช้ `.get()` จะช่วยป้องกันการเกิด `KeyError` เมื่อค้นหา Key ที่ไม่มีอยู่ในระบบ

```python
# การเข้าถึงข้อมูลแบบปลอดภัยด้วย .get()
weather_data = {"temp": 24}

# Indexing ปกติ (อันตราย): v = weather_data["humid"] จะทำให้เกิด ERROR
# ใช้ .get() (ปลอดภัย):
humidity = weather_data.get("humid", 0) # หากไม่พบ Key "humid" จะคืนค่า 0
print(f"Humidity: {humidity}")

```

### Code Snippet: การรวม Dictionary (Modern Merge Operators)

ตั้งแต่ Python 3.9 ขึ้นไป คุณสามารถใช้เครื่องหมาย `|` เพื่อรวม Dictionary ได้อย่างสะอาดตา

```python
d1 = {'a': 1, 'b': 2}
d2 = {'b': 3, 'c': 4}

# Union Operator: สร้าง Dictionary ใหม่ (ค่าจากตัวขวาจะทับตัวซ้ายหาก Key ซ้ำกัน)
merged = d1 | d2
# ผลลัพธ์: {'a': 1, 'b': 3, 'c': 4}

# In-Place Update: อัปเดตค่าไปยัง d1 โดยตรง (คล้ายคำสั่ง d1.update(d2))
d1 |= d2

```

### Code Snippet: Dictionary Comprehensions และ ZIP()

การใช้ `zip()` มีประโยชน์มากในการจับคู่ List สองตัวเข้าด้วยกันเพื่อสร้างเป็น Dictionary

```python
# การใช้ zip() จับคู่ข้อมูล
fields = ("name", "age", "job")
data = ["Dev", 28, "Engineer"]
profile = dict(zip(fields, data))
# ผลลัพธ์: {'name': 'Dev', 'age': 28, 'job': 'Engineer'}

# Dictionary Comprehension: การสลับตำแหน่ง Key และ Value
d = {"A": 1, "B": 2}
inv = {v: k for k, v in d.items()}
# ผลลัพธ์: {1: 'A', 2: 'B'}

```

## 2. Sets (The Unique Collection)

Set คือคอลเลกชันของข้อมูลที่ไม่ซ้ำกัน (Unique elements) และไม่มีการเรียงลำดับ (Unordered)

* Set เหมาะสำหรับการลบข้อมูลที่ซ้ำกันออกจาก List และการตรวจสอบการเป็นสมาชิกของข้อมูลอย่างรวดเร็ว


* สามารถใช้การดำเนินการทางคณิตศาสตร์เกี่ยวกับเซตได้ เช่น Union, Intersection, Difference


* **ข้อควรระวัง:** การสร้าง Set ว่างต้องใช้คำสั่ง `set()` เท่านั้น เพราะการใช้ `{}` จะถือเป็นการสร้าง Dictionary ว่าง



### Code Snippet: การดำเนินการทางคณิตศาสตร์ของ Set (Mathematical Operations)

```python
# การหาข้อมูลนักเรียนที่ลงทะเบียนเรียน
python_students = {"S001", "S003", "S005"}
java_students = {"S002", "S003", "S006"}

# Intersection (&): หาคนที่เรียนทั้งสองวิชา (ข้อมูลที่ซ้ำกัน)
both_courses = python_students & java_students
print(f"Students in both: {both_courses}") # ผลลัพธ์: {'S003'}

# Union (|): หาคนทั้งหมดจากทั้งสองวิชา (รวมกันแบบไม่ซ้ำ)
all_unique_students = python_students | java_students

# Difference (-): หาคนที่เรียน Python แต่ไม่เรียน Java
only_python = python_students - java_students

```

## 3. สรุปเปรียบเทียบโครงสร้างข้อมูล (Sequence vs Mapping)

การเลือกใช้ Data Structure ให้เหมาะสมกับงานเป็นสิ่งสำคัญ:

* **List (Sequence):** เปลี่ยนแปลงค่าได้ (Mutable), ใช้ Index เป็นตัวเลข, ความเร็วการค้นหา O(n) Linear Search, เหมาะสำหรับข้อมูลที่ต้องการเก็บลำดับ


* **Tuple (Constant):** เปลี่ยนแปลงค่าไม่ได้ (Immutable), ใช้ Index เป็นตัวเลข, ความเร็วการค้นหา O(n) Linear Search, เหมาะสำหรับระเบียนข้อมูลที่ตายตัว


* **Dict (Mapping):** เปลี่ยนแปลงค่าได้ (Mutable), ใช้ Key เป็นตัวอ้างอิง, ความเร็วการค้นหา O(1) Constant Hash, เหมาะสำหรับข้อมูลที่มีป้ายกำกับ (Labeled Entities)

## 4. โบนัส: The Collections Module

สำหรับงานที่เฉพาะเจาะจง Python มีโมดูล `collections` ที่ทรงพลังเพื่อลดภาระในการเขียนโค้ด:

* `DefaultDict`: สร้างค่าเริ่มต้นให้กับ Key ที่ยังไม่มีอยู่ใน Dictionary อัตโนมัติ


* `Counter`: เป็น Dictionary ที่ถูกปรับแต่งมาเพื่อใช้สำหรับการนับจำนวนวัตถุโดยเฉพาะ



```python
from collections import defaultdict, Counter

# การใช้ defaultdict
d = defaultdict(int)
d["clicks"] += 1 # ไม่เกิด Error แม้ "clicks" จะยังไม่มีอยู่ก่อนหน้า

# การใช้ Counter เพื่อนับความถี่
c = Counter("banana")
# ผลลัพธ์: Counter({'a': 3, 'n': 2, 'b': 1})

```

# ประโยชน์หลักของ Set

* **การกำจัดข้อมูลที่ซ้ำกัน (Removing Duplicates):** เนื่องจาก Set เป็นคอลเลกชันที่เก็บเฉพาะข้อมูลที่ไม่ซ้ำกัน (Unique elements) ดังนั้นหากคุณมีกลุ่มข้อมูลที่มีค่าซ้ำซ้อนกัน การแปลงข้อมูลเหล่านั้นให้เป็น Set จะเป็นการลบตัวซ้ำออกไปโดยอัตโนมัติ


* **การตรวจสอบสถานะการเป็นสมาชิกอย่างรวดเร็ว (Efficient Membership Testing):** Set ถูกออกแบบมาให้มีประสิทธิภาพสูงมากในการตรวจสอบว่ามีข้อมูลตัวใดตัวหนึ่งอยู่ในกลุ่มข้อมูลนั้นหรือไม่ โดยไม่ต้องสนใจเรื่องลำดับการจัดเก็บ


* **การดำเนินการตามทฤษฎีเซตทางคณิตศาสตร์ (Mathematical Set Theory):** คุณสามารถจัดการข้อมูลระหว่างกลุ่มได้อย่างง่ายดาย เช่น การรวมข้อมูลทั้งหมด (Union), การหาข้อมูลที่เหมือนกัน (Intersection) หรือการหาข้อมูลที่แตกต่างกัน (Difference)



### ตัวอย่างการประยุกต์ใช้งานในสถานการณ์จริง

* การเปรียบเทียบข้อมูลสองกลุ่ม เช่น การค้นหานักเรียนที่ลงทะเบียนเรียนซ้ำซ้อนใน 2 รายวิชา
