# Activity 4.1: List vs. Tuple: Scenario Matching

##สิ่งที่อยากให้ทำ : สถานการณ์ต่อไปนี้ แต่ละเหตุการณ์ควรใช้ list หรือ tuple เพราะอะไร


###1ข้อมูลอุณหภูมิประจำวันที่มีการบันทึกไว้ตลอดหนึ่งสัปดาห์

ตอบ ควรใช้ list เพราะอุณหภูมิเป็นข้อมูลที่มีการเปลี่ยนแปลงตลอดที่บันทึกไว้ การดำเนินการด้วย list จะสามารถเพิ่มข้อมูลได้ง่ายกว่าและมี 1 มิติ


###2.ค่าสี RGB ของเฉดสีเฉพาะเฉดหนึ่ง (เช่น สีแดงคือ (255, 0, 0))

ตอบ tuple เพราะ ค่าสี RGB ของเฉดสีเฉพาะเฉดหนึ่ง ไม่เปลี่ยนแปลง โดยข้อดีที่ใช้ tuple คือจะประมวลผลเร็วกว่า list และไม่สามารถแก้ไขค่าได้จึงเหมะกับสถานการณ์นี้


###3รายการสินค้าในรถเข็นชอปปิงที่สามารถกดเพิ่มหรือลบออกได้

ตอบ List เพราะ list สามารถจัดการข้อมูลโดยการลบหรือเพิ่มได้ดี และมี 1 มิติ

###4.รายชื่อวันในหนึ่งสัปดาห์เรียงตามลำดับ (จันทร์ - อาทิตย์)

ตอบ tuple เพราะ รายชื่อวันใน 1 สัปดาห์จะไม่เปลี่ยนแปลง จึงใช้ tuple ที่ไม่สามารถแก้ไขค่าได้


###5.คะแนนเก็บของนักเรียนจากการส่งงานชุดต่างๆ ซึ่งอาจมีการอัปเดตหรือแก้ไขคะแนนในภายหลังได้

ตอบ List เพราะสามารถแก้ไขคะแนนภายหลังได้ ไม่เหมือน tuple


###6.พิกัด GPS ของสถานที่สำคัญหรือจุดแลนด์มาร์กที่ตั้งอยู่กับที่

ตอบ tuple เพราะ พิกัด GPS ของสถานที่สำคัญหรือจุดแลนด์มาร์กที่ตั้งอยู่กับที่ จะไม่สามารถเปลี่ยนแปลงได้ โดยข้อดีที่ใช้ tuple คือจะประมวลผลเร็วกว่า list และไม่สามารถแก้ไขค่าได้จึงเหมะกับสถานการณ์นี้
