# Recap Week 4: Lists และ Tuples (5-Minute Recap)

**1. List (ลิสต์): โครงสร้างข้อมูลแบบแก้ไขได้ (Mutable)**

* **ลักษณะสำคัญ:** เป็นข้อมูลแบบลำดับ (Ordered) สามารถเปลี่ยนแปลงค่าได้ (Mutable) และอนุญาตให้มีข้อมูลซ้ำกันได้


* **การสร้างและการเข้าถึง:** สร้างโดยใช้เครื่องหมายวงเล็บเหลี่ยม `[]` สามารถเก็บข้อมูลต่างชนิดกันรวมไว้ในลิสต์เดียวกัน หรือสร้างลิสต์ซ้อนลิสต์ (Nested list) ได้ การเข้าถึงข้อมูลสามารถใช้การระบุตำแหน่ง (Indexing) และการตัดแบ่ง (Slicing) เช่น `list[start:end:step]`


* **เมธอดในการจัดการข้อมูล (List Methods):**
* **เพิ่มข้อมูล:** `append(item)` นำไปต่อท้าย, `insert(index, item)` แทรกตามตำแหน่งที่ระบุ, `extend(another_list)` นำสมาชิกจาก iterable อื่นมาต่อท้าย


* **ลบข้อมูล:** `remove(item)` ลบค่าที่ตรงกันตัวแรก, `pop(index)` ดึงและลบค่าตามตำแหน่ง, `del list[index]` ลบตามตำแหน่ง, `clear()` ลบข้อมูลทั้งหมด


* **การจัดเรียงและการค้นหา:** `sort()` เรียงลำดับข้อมูลในลิสต์เดิม, `reverse()` กลับลำดับข้อมูล, `count(item)` นับจำนวนข้อมูลที่ซ้ำกัน, `index(item)` หาตำแหน่งแรกของข้อมูลนั้น




* **ข้อควรระวังในการคัดลอก (Copying):** การใช้เครื่องหมาย `=` เป็นการคัดลอกเพียงจุดอ้างอิง (Reference) ไปยังออบเจกต์เดียวกัน หากมีการแก้ไขจะส่งผลกระทบต่อกัน ดังนั้นควรใช้การ Slicing (`new_list = old_list[:]`) หรือฟังก์ชัน `list()` เพื่อสร้างชุดข้อมูลใหม่ที่แยกขาดจากกัน



```python
# ตัวอย่าง List สำหรับบันทึกประวัติการเรียนรู้ (Learning Log)
learning_log = ["Variables", "Loops", "Conditionals"]

# การเพิ่มและแทรกข้อมูล
learning_log.append("Lists")
learning_log.insert(1, "Data Types")
print(learning_log)
# Output: ['Variables', 'Data Types', 'Loops', 'Conditionals', 'Lists']

# การลบข้อมูลและเรียกดูจำนวน
learning_log.remove("Loops")
print(f"Total topics learned: {len(learning_log)}")
# Output: Total topics learned: 4

```

**2. Tuple (ทูเพิล): โครงสร้างข้อมูลแบบแก้ไขไม่ได้ (Immutable)**

* **ลักษณะสำคัญ:** เป็นข้อมูลแบบลำดับ (Ordered) ที่ **ไม่สามารถเปลี่ยนแปลงค่าได้ (Immutable)** หลังจากถูกสร้างขึ้น แต่ยังอนุญาตให้มีข้อมูลซ้ำกันได้


* **การสร้าง:** ใช้เครื่องหมายวงเล็บกลม `()` ในการสร้าง


* **ไวยากรณ์เฉพาะ:** หากต้องการสร้าง Tuple ที่มีสมาชิกเพียงตัวเดียว จะต้องใส่เครื่องหมายจุลภาคต่อท้ายเสมอ เช่น `("apple",)` ไม่เช่นนั้นระบบจะมองว่าเป็นเพียงค่าสตริงหรือตัวเลขธรรมดา


* **ข้อจำกัด (Immutability):** หากมีความพยายามที่จะแก้ไข เพิ่ม หรือลบสมาชิกภายใน Tuple โปรแกรมจะแจ้งข้อผิดพลาดเป็น `TypeError` หรือ `AttributeError` ทันที


* **ข้อดีและสถานการณ์ที่ควรใช้:** เหมาะสำหรับการเก็บข้อมูลที่คงที่และไม่ควรถูกแก้ไข (เช่น พิกัดภูมิศาสตร์ หรือข้อมูลการตั้งค่า) สามารถนำไปใช้เป็นคีย์ใน Dictionary ได้เนื่องจากไม่สามารถแก้ไขได้ และใช้พื้นที่หน่วยความจำน้อยกว่า List เมื่อเทียบในปริมาณข้อมูลที่เท่ากัน



```python
# ตัวอย่าง Tuple สำหรับข้อมูลที่ไม่มีการเปลี่ยนแปลง
course_details = ("Python Fundamentals", "Week 4", 2026)

# ตัวอย่างข้อผิดพลาดเมื่อพยายามแก้ไข Tuple
# course_details[1] = "Week 5"  # จะเกิด TypeError

# การแปลงชนิดข้อมูล (Type Conversion)
log_tuple = tuple(learning_log) # แปลง List เป็น Tuple[cite: 3]
log_list = list(course_details) # แปลง Tuple กลับเป็น List[cite: 3]

```
