#### 1. การสร้างและใช้งานฟังก์ชัน (Defining & Calling Functions)

* **โครงสร้าง:** ใช้คีย์เวิร์ด `def` ตามด้วยชื่อฟังก์ชัน และ Parameter ในวงเล็บ


* **การส่งค่ากลับ:** ใช้ `return` เพื่อคืนค่ากลับไปยังผู้เรียก หากไม่มีการสั่ง `return` ฟังก์ชันจะคืนค่า `None` โดยอัตโนมัติ


* **การคืนหลายค่า:** Python สามารถคืนค่าหลายค่าพร้อมกันได้ โดยระบบจะแพ็กค่านั้นเป็น `tuple` ให้อัตโนมัติ



#### 2. Arguments & Parameters

* **Positional vs Keyword:** เรียงตามลำดับตำแหน่ง หรือระบุชื่อ `key=value` เพื่อความชัดเจน


* **Default Values:** กำหนดค่าเริ่มต้นให้ Parameter ได้ **ข้อควรระวัง:** ห้ามใช้ Mutable Objects (เช่น `list`, `dict`) เป็นค่าเริ่มต้น ให้ใช้ `None` แทนเพื่อป้องกันบั๊ก


* **Arbitrary Arguments:**
* `*args`: รับ Positional Arguments ไม่จำกัดจำนวน (แปลงเป็น `tuple`)


* `**kwargs`: รับ Keyword Arguments ไม่จำกัดจำนวน (แปลงเป็น `dict`)





#### 3. ขอบเขตตัวแปร (Variable Scope) & LEGB Rule

* **Local vs Global:** ตัวแปรที่สร้างในฟังก์ชันถือเป็น Local Variable ไม่สามารถเรียกใช้นอกฟังก์ชันได้


* **`global` Keyword:** หากต้องการแก้ไขตัวแปร Global จากภายในฟังก์ชัน ต้องระบุด้วยคำสั่ง `global`

* **การค้นหาตัวแปร (LEGB):** Local $\rightarrow$ Enclosing $\rightarrow$ Global $\rightarrow$ Built-in



#### 4. โมดูลและการใช้งาน (Modules & Imports)

* โมดูลคือไฟล์ `.py` ที่เก็บรวบรวมฟังก์ชันและตัวแปรไว้เพื่อเรียกใช้ซ้ำ


* **วิธีนำเข้า:** `import module_name`, `import module_name as alias`, หรือ `from module_name import function_name`

* **Standard Libraries:** โมดูลติดเครื่องที่นิยมใช้ เช่น `math` และ `random`

#### Snippet 1: การนิยามฟังก์ชัน, Mutable Defaults และ Scope

```python
# 1. การรับค่าแบบ Dynamic (*args, **kwargs) และ Multiple Return Values
def calculate_stats(*numbers):
    total = sum(numbers)
    avg = total / len(numbers) if numbers else 0
    return total, avg  # คืนค่ากลับมาเป็น Tuple

# Unpacking ค่าที่ส่งกลับมา
sum_val, avg_val = calculate_stats(10, 20, 30, 40)
print(f"Sum: {sum_val}, Avg: {avg_val}")  # Sum: 100, Avg: 25.0

# 2. แนวทางปฏิบัติที่ถูกต้องสำหรับ Mutable Default Arguments
def add_item(item, target_list=None):
    if target_list is None:
        target_list = []  # ป้องกันการแชร์ List ร่วมกันระหว่างการเรียกฟังก์ชัน
    target_list.append(item)
    return target_list

# 3. ขอบเขตตัวแปร (Variable Scope) และการใช้คำสั่ง global
counter = 0

def increment():
    global counter
    counter += 1

increment()
print(f"Global Counter: {counter}")  # Output: 1

```

#### Snippet 2: Unpacking Arguments และการใช้งาน Modules

```python
import math
import random as rnd

# 1. Dictionary & List Unpacking ในฟังก์ชัน
def display_profile(name, age, position="Developer"):
    print(f"Name: {name}, Age: {age}, Position: {position}")

user_info = {"name": "Alice", "age": 25}
display_profile(**user_info)  # Unpacking dict เข้าไปใน Keyword arguments

# 2. การสร้าง custom module บน Google Colab ด้วย %%writefile
# (บรรทัดนี้ใช้ใน Colab เพื่อสร้างไฟล์ my_utils.py)

```

```python
%%writefile my_utils.py
def is_even(num):
    return num % 2 == 0

```

```python
# นำเข้า Custom Module และ Standard Library มาใช้งาน
import my_utils

print(f"Is 10 Even?: {my_utils.is_even(10)}")
print(f"Square root of 16: {math.sqrt(16)}")
print(f"Random Roll (1-6): {rnd.randint(1, 6)}")

```
