# Lab 6.2: Custom Module and Standard Library Usage

```python
%%writefile my_utils.py

def greet(name):
    print(f"Hello, {name}!")


def is_prime(number):
    if number < 2:
        return False

    for i in range(2, number):
        if number % i == 0:
            return False

    return True
```

```python
%%writefile main_program.py

import my_utils
import math
import random

my_utils.greet("Alice")

print("Is 7 prime?", my_utils.is_prime(7))
print("Is 10 prime?", my_utils.is_prime(10))

print("Square root of 25 =", math.sqrt(25))

random_number = random.randint(1, 100)
print("Random number =", random_number)
```
