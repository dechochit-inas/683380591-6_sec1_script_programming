#Lab 6.1: Simple Calculator Functions

```python
def add(a , b):
  print(f"{a} + {b} = {a+b}")
  return a+b
def subtract(a,b):
  print(f"{a} - {b} = {a-b}")
  return a-b
def multiply(a,b):
  print(f"{a} * {b} = {a*b}")
  return a*b
def divide(a,b):
  if b == 0:
    print("Cannot divide by zero")
    return
  print(f"{a} / {b} = {a/b}")
  return a/b


while True:
  print("Menu\n1. add\n2. subtract\n3. multiply\n4. divide\n5. exit")
  choice = int(input("Enter your choice: "))
  if choice == 1:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    add(a,b)
  elif choice == 2:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    subtract(a,b)
  elif choice == 3:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    multiply(a,b)
  elif choice == 4:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    divide(a,b)
  elif choice == 5:
    break
  else:
    print("Invalid choice")
```
